```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('student_data.csv')
```

```python
df.columns
```

```python
plt.hist(df['age'], bins=10, edgecolor='black')
plt.xlabel('Age')
plt.ylabel('Number of Students')
plt.title('Distribution of Student Age')
plt.show()
```

```python
plt.figure(figsize=(8,5))
plt.hist(df['G3'], bins=10, edgecolor='black')
plt.xlabel('Final Grade (G3)')
plt.ylabel('Number of Students')
plt.title('Distribution of Final Grades')
plt.show()
```

```python
plt.figure(figsize=(6,5))
plt.boxplot(df['G3'])
plt.ylabel('Final Grade (G3)')
plt.title('Boxplot of Final Grades')
plt.show()
```

```python
plt.scatter(df['G1'], df['G3'])
plt.xlabel('G1')
plt.ylabel('G3')
plt.title('G1 vs Final Grade (G3)')
plt.show()
```

```python
df['sex'].value_counts().plot(kind='bar')
plt.xlabel('Gender')
plt.ylabel('Number of Students')
plt.title('Number of Students by Gender')
plt.show()
```

```python
numeric_columns = df.select_dtypes(include='number').columns

plt.figure(figsize=(15, 8))
df[numeric_columns].boxplot()
plt.xticks(rotation=45)
plt.title('Boxplots of Numeric Columns')
plt.show()
```

```python
df.hist(figsize=(15, 15), bins=10, edgecolor='black')
plt.tight_layout()
plt.show()
```
